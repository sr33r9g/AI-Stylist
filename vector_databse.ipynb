```python
#chorma db implementaion (vector database creation)
import chromadb
import json
```

```python
client=chromadb.PersistentClient('./my_local_data') #data is save to folder
```

```python
collection=client.get_or_create_collection(name='my_knowledge_base') # creating collection
```

```python
with open('sample_data.json','r') as file:
    style_data=json.load(file)
```

```python
document=[]
metadata=[]
ids=[]

for item in style_data:
    content=(
        f"Occasion: {item['sub_occasion']} for {item['occasion']}. "
        f"Guidelines: {item['rule']} "
        f"Color Strategy: {item['color_logic']} "
        f"Recommended Accessories: {item['accessory']}"
    )
    document.append(content)
    metadata.append({
        "occasion": item["occasion"],
        "sub_occasion": item["sub_occasion"],
        "id_tag": item["id"]
    })
    ids.append(item["id"])
```

```python
collection.add(documents=document,metadatas=metadata,ids=ids)
```

```python
results = collection.query(
    query_texts=["User is wearing a slim tight fit white trousers made of linen,User is wearing a regular standard fit blue long_sleeved_shirt made of denim rugged material."],
    n_results=3
)
```
